### Import Libraries and Download Data

```python
# Import required NLP libraries
import pandas as pd
import spacy
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

# Download NLTK datasets for tokenization and stopwords quietly
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

# Load the spaCy English language model
nlp = spacy.load("en_core_web_sm")
print("NLP Libraries loaded successfully!")
```

### Defining a Sample Text

```python
# Define the test sentence
sample_text = "We are looking for candidates with strong experience in Python and SQL."
print("Original Text:", sample_text)
```

### Sentence Segmentation and Lowercasing

```python
# Process the text with spaCy (lowercasing it first)
doc = nlp(sample_text.lower())

# Extract sentences
print("Sentence Segmentation:")
for i, sent in enumerate(doc.sents):
    print(f" Sentence {i+1}: {sent.text}")
```

### Tokenization

```python
# Extract individual tokens
tokens = [token.text for token in doc]

print("Tokenized:")
print(" | ".join(tokens))
```

### StopWord Removal

```python
# Load standard English stopwords
stop_words = set(stopwords.words('english'))

# Filter out stopwords and non-alphabetical tokens (like punctuation)
filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]

print("After Stop Word Removal:")
print(" | ".join(filtered_tokens))
```

### Comparing Lemmatization and Stemming

```python
# 1. Stemming using NLTK
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

# 2. Lemmatization using spaCy
lemmatized_words = [token.lemma_ for token in doc if token.text in filtered_tokens]

print("Stemming (NLTK):")
print(" | ".join(stemmed_words))

print("\nLemmatization (spaCy):")
print(" | ".join(lemmatized_words))
```
