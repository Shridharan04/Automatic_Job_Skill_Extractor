# Day 10 : N-Gram Analysis

### Importing and Loading Data 

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

print("Loading clean_jobs.csv...")
df = pd.read_csv("clean_jobs.csv")

# Ensuring all text is treated as strings to avoid vectorizer errors
df['clean_description'] = df['clean_description'].fillna("")
```

### Initializing the N-Gram Vectorizer

```python
# Setting ngram_range=(1,3) to capture single words, pairs, and triplets
ngram_vec = TfidfVectorizer(
    ngram_range=(1, 3), 
    max_features=1500, 
    stop_words="english"
)

print("Applying N-Gram Vectorizer to the dataset (this may take a moment)...")
X_ngram = ngram_vec.fit_transform(df["clean_description"])
print("Vectorization complete!")
```

### Extracting and Sorting Terms

```python
# Get the actual words/phrases generated
ngram_features = ngram_vec.get_feature_names_out()

# Calculate the overall score for each phrase across the entire dataset
ngram_scores = X_ngram.sum(axis=0).A1

# Create a DataFrame to view the results cleanly
ngram_df = pd.DataFrame({'term': ngram_features, 'score': ngram_scores})

print("Extraction complete. DataFrame created.")
```

### Discovering Multi-Word Skills 

```python
# Filter specifically for multi-word phrases (checking for the space character)
multi_word_phrases = ngram_df[ngram_df['term'].str.contains(" ")]

print("Top Multi-Word Phrases Discovered:\n")
# Display the top 20 to see if it successfully caught technical skills
display(multi_word_phrases.sort_values(by='score', ascending=False).head(20))
```

# Conclusion : Multi-Word Skill Discovery

By expanding the TF-IDF vectorizer to include bigrams and trigrams, the model successfully identified crucial multi-word phrases that a single-word tokenizer would miss. This approach is highly effective for discovering complex technical skills like "Machine Learning," "Data Science," and "Computer Vision" embedded within the job descriptions.
