# Term Frequency - Inverse Document Frequency (TF-IDF)

### Importing Libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

### Loading Data 

```python
df = pd.read_csv("clean_jobs.csv")
df['clean_description'] = df['clean_description'].fillna("")
```

### Initialize TF-IDF Vectorizer

```python
vectorizer = TfidfVectorizer(max_features=1000, stop_words="english")
X = vectorizer.fit_transform(df["clean_description"])
```

### Extracting terms and their overall scores

```python
feature_names = vectorizer.get_feature_names_out()
scores = X.sum(axis=0).A1
tfidf_df = pd.DataFrame({'term': feature_names, 'score': scores})
```

### Displaying the highest-scoring terms

```python
print("Top TF-IDF Terms:")
display(tfidf_df.sort_values(by='score', ascending=False).head(15))
```

### Conclusion : Does  TF-IDF identify useful skills autoatically ?

No, it only partially succeeds. As seen in our output, TF-IDF successfully identifies highly frequent, unique words without needing a hardcoded dictionary. However, because it lacks contextual awareness of what a technical "skill" actually is, it highly ranks generic corporate buzzwords (like "team", "work", "sales", and "business") right alongside actual technical requirements. It is a powerful statistical tool, but not a precise skill extractor on its own.
