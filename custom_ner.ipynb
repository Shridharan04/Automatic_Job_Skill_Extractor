# Day 12 - Named Entity Recognition

### The Problem with Standard NER

```python
import spacy

# Load the standard English model
nlp = spacy.load("en_core_web_sm")

# A standard technical requirement sentence
text = "We are looking for experts in Python, AWS, and React based in New York."
doc = nlp(text)

print("Standard NER Output:")
print("-" * 30)
for ent in doc.ents:
    print(f"Word: {ent.text:15} | Entity Type: {ent.label_}")
```

### Building the Custom Tech Ruler

```python
# Create a fresh pipeline to avoid conflicts
nlp_custom = spacy.load("en_core_web_sm")

# Add an EntityRuler BEFORE the standard NER model processes the text
ruler = nlp_custom.add_pipe("entity_ruler", before="ner")

# Define our strict custom categories
patterns = [
    {"label": "SKILL", "pattern": "Python"},
    {"label": "SKILL", "pattern": "Machine Learning"},
    {"label": "CLOUD_PLATFORM", "pattern": "AWS"},
    {"label": "CLOUD_PLATFORM", "pattern": "Google Cloud"},
    {"label": "DATABASE", "pattern": "PostgreSQL"},
    {"label": "DATABASE", "pattern": "MongoDB"},
    {"label": "TOOL", "pattern": "Power BI"},
    {"label": "TOOL", "pattern": "Tableau"}
]

# Inject the patterns into the AI model
ruler.add_patterns(patterns)
print("Custom technical entities successfully added to the NLP pipeline!")
```

### Testing the Upgraded Recruiter Tool

```python
jd_text = """
The ideal candidate will have strong Python programming skills and a deep understanding 
of Machine Learning. You must be comfortable managing data in PostgreSQL and MongoDB, 
deploying pipelines on AWS or Google Cloud, and building dashboards using Power BI.
"""

doc_custom = nlp_custom(jd_text)

print("\nCustom Tech NER Output:")
print("-" * 30)
for ent in doc_custom.ents:
    # Filter to only show our custom technical tags and locations
    if ent.label_ in ["SKILL", "CLOUD_PLATFORM", "DATABASE", "TOOL", "GPE"]:
        print(f"Word: {ent.text:20} | Category: {ent.label_}")
```

### Advanced NER stress tests

```python
# --- ADVANCED NER STRESS TESTS ---

print("Expanding taxonomy and initiating advanced NER tests...\n" + "="*50)

# 1. Inject a few more modern skills into our existing ruler
advanced_patterns = [
    {"label": "SKILL", "pattern": "React"},
    {"label": "SKILL", "pattern": "Node.js"},
    {"label": "FRAMEWORK", "pattern": "Express"},
    {"label": "TOOL", "pattern": "Docker"},
    {"label": "TOOL", "pattern": "Kubernetes"}
]
ruler.add_patterns(advanced_patterns)

# 2. Define Real-World Recruiter Edge Cases
test_cases = {
    "The Full-Stack Web Developer": "Seeking a developer to build interactive frontends with React and scalable backend APIs using Node.js, Express, and MongoDB.",
    "The Cloud & DevOps Engineer": "Need an expert to containerize microservices using Docker, orchestrate with Kubernetes, and deploy exclusively on AWS.",
    "The Tricky Mixed JD": "Looking for a Data Scientist in San Francisco. Must know Python and Machine Learning, but bonus points if you can build a quick dashboard in Tableau."
}

# 3. Run the engine against all test cases
for role, description in test_cases.items():
    print(f"\nScenario: {role}")
    print(f"Text: '{description}'")
    print("-" * 30)
    
    doc = nlp_custom(description)
    found_entities = False
    
    for ent in doc.ents:
        if ent.label_ in ["SKILL", "FRAMEWORK", "CLOUD_PLATFORM", "DATABASE", "TOOL", "GPE"]:
            print(f"Detected: {ent.text:15} | Category: {ent.label_}")
            found_entities = True
            
    if not found_entities:
        print("No custom technical entities found.")
```

### Conclusion : Why does standard NER fail to recognize many technical skills?

Standard NER models (like spaCy's en_core_web_sm) are trained primarily on standard web text and news articles. Their primary goal is identifying generic entities like People (PERSON), Organizations (ORG), and Locations (GPE). Because they lack specialized training on engineering documentation, they fail to recognize tech skills, often missing the context of software tools entirely. By adding a custom entity_ruler, we force the model to explicitly map specific terminology to our custom technical taxonomy, allowing for precise, categorized extraction.
